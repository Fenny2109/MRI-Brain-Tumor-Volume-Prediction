## 10. Multicollinearity Check Using VIF

VIF is calculated using the final predictor set. Very high VIF values suggest redundant predictors.

```python
X_vif = sm.add_constant(X.copy())

vif_data = pd.DataFrame({
    "Feature": X_vif.columns,
    "VIF": [
        variance_inflation_factor(
            X_vif.values,
            i
        )
        for i in range(X_vif.shape[1])
    ]
})

vif_data = vif_data[
    vif_data["Feature"] != "const"
].sort_values(
    "VIF",
    ascending=False
)

display(vif_data)
```

## 11. Train/Test Split

The same split is used for all regression models for fair comparison.

```python

X_train , X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

## 12. LASSO Feature Selection

```python
lasso_fs = Pipeline([
    ("scaler", StandardScaler()),
    ("lasso", LassoCV(cv=5, max_iter=100000, random_state=42))
])

lasso_fs.fit(X_train, y_train)

coef = pd.Series(
    lasso_fs.named_steps["lasso"].coef_,
    index=X_train.columns
)

selected = coef[coef != 0].sort_values(key=abs, ascending=False)
selected
```

## 13. Linear Regression

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

pred_lr = lr.predict(X_test)

mae_lr = mean_absolute_error(y_test, pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, pred_lr))
r2_lr = r2_score(y_test, pred_lr)

print("Linear Regression")
print("MAE:", mae_lr)
print("RMSE:", rmse_lr)
print("R2:", r2_lr)
```

## 14. Ridge Regression

```python
alphas = [0.01, 0.1, 1, 10, 100]

ridge = Pipeline([
    ("scaler", StandardScaler()),
    ("ridge", RidgeCV(alphas=alphas))
])

ridge.fit(X_train, y_train)
pred_ridge = ridge.predict(X_test)

mae_ridge = mean_absolute_error(y_test, pred_ridge)
rmse_ridge = np.sqrt(mean_squared_error(y_test, pred_ridge))
r2_ridge = r2_score(y_test, pred_ridge)

print("Ridge Regression")
print("Best Alpha:", ridge.named_steps["ridge"].alpha_)
print("MAE:", mae_ridge)
print("RMSE:", rmse_ridge)
print("R2:", r2_ridge)
```

## 15. LASSO Regression Model

This uses LASSO as a prediction model, not only as feature selection.

```python
lasso_model = Pipeline([
    ("scaler", StandardScaler()),
    ("lasso", LassoCV(cv=5, max_iter=100000, random_state=42))
])

lasso_model.fit(X_train, y_train)
pred_lasso = lasso_model.predict(X_test)

mae_lasso = mean_absolute_error(y_test, pred_lasso)
rmse_lasso = np.sqrt(mean_squared_error(y_test, pred_lasso))
r2_lasso = r2_score(y_test, pred_lasso)

print("LASSO Regression")
print("MAE:", mae_lasso)
print("RMSE:", rmse_lasso)
print("R2:", r2_lasso)
```

## Elastic Net Regression

```python
elastic_net = make_pipeline(
    StandardScaler(),
    ElasticNetCV(
        l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1.0],
        cv=10,
        max_iter=10000,
        random_state=42
    )
)

elastic_net.fit(X_train, y_train)

# Numeric regression predictions
pred_elastic = elastic_net.predict(X_test)

elastic_model = elastic_net.named_steps["elasticnetcv"]

mae_elastic = mean_absolute_error(y_test, pred_elastic)
rmse_elastic = np.sqrt(mean_squared_error(y_test, pred_elastic))
r2_elastic = r2_score(y_test, pred_elastic)

print("Elastic Net Regression")
print(f"Best alpha    : {elastic_model.alpha_:.4f}")
print(f"Best l1_ratio : {elastic_model.l1_ratio_}")
print(f"Train R²      : {elastic_net.score(X_train, y_train):.4f}")
print(f"Test R²       : {r2_elastic:.4f}")
print(f"Test RMSE     : {rmse_elastic:.3f}")
print(f"Test MAE      : {mae_elastic:.3f}")
```

## 21. Clustering 

```python
# ============================================================
# K-MEANS CLUSTERING
# ============================================================

# Scale predictors
Xs = StandardScaler().fit_transform(X)

# Compare different values of k
k_results = []

for k in range(2, 10):

    km_test = KMeans(
        n_clusters=k,
        n_init=20,
        random_state=42,
        algorithm="lloyd"
    )

    cluster_labels = km_test.fit_predict(Xs)

    silhouette = silhouette_score(
        Xs,
        cluster_labels
    )

    k_results.append({
        "k": k,
        "Inertia": km_test.inertia_,
        "Silhouette": silhouette
    })

    print(
        f"k={k}  "
        f"inertia={km_test.inertia_:8.1f}  "
        f"silhouette={silhouette:.4f}"
    )

k_results = pd.DataFrame(k_results)

display(k_results)
```

```python
# Final K-Means model
best_k = 3

km_final = KMeans(
    n_clusters=best_k,
    n_init=20,
    random_state=42,
    algorithm="lloyd"
)

cluster_labels = km_final.fit_predict(Xs)

X_clustered = X.copy()
X_clustered["cluster"] = cluster_labels

print("\nCluster means:")
display(
    X_clustered.groupby("cluster").mean().round(2)
)

print("\nCluster sizes:")
print(
    X_clustered["cluster"].value_counts().sort_index()
)
```

```python
# PCA for visualization
pca_cluster = PCA(n_components=2)

pcs = pca_cluster.fit_transform(Xs)

centroids_pca = pca_cluster.transform(
    km_final.cluster_centers_
)

variance = (
    pca_cluster.explained_variance_ratio_ * 100
)

plt.figure(figsize=(8, 6))

for cluster_number in range(best_k):

    mask = cluster_labels == cluster_number

    plt.scatter(
        pcs[mask, 0],
        pcs[mask, 1],
        s=50,
        alpha=0.75,
        label=(
            f"Cluster {cluster_number} "
            f"(n={mask.sum()})"
        )
    )

plt.scatter(
    centroids_pca[:, 0],
    centroids_pca[:, 1],
    marker="X",
    s=250,
    label="Centroids"
)

plt.xlabel(
    f"PC1 ({variance[0]:.1f}%)"
)

plt.ylabel(
    f"PC2 ({variance[1]:.1f}%)"
)

plt.title(
    f"K-Means Clustering with k={best_k}"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Logistic Regression Classification 

```python


# Split the classification data
Xtr, Xte, ytr, yte = train_test_split(
    X,
    y_class,
    test_size=0.25,
    random_state=42,
    stratify=y_class
)

# Create and train Logistic Regression model
clf = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=2000,
        random_state=42
    )
)

clf.fit(Xtr, ytr)

# Predictions
pred_logistic = clf.predict(Xte)

# Cross-validation accuracy
cv_accuracy = cross_val_score(
    clf,
    Xtr,
    ytr,
    cv=10,
    scoring="accuracy"
).mean()

# Test accuracy
test_accuracy = accuracy_score(
    yte,
    pred_logistic
)

# Majority-class baseline
baseline_accuracy = y_class.value_counts(
    normalize=True
).max()

print("Logistic Regression Classification")
print(f"CV accuracy   : {cv_accuracy:.4f}")
print(f"Test accuracy : {test_accuracy:.4f}")
print(f"Baseline      : {baseline_accuracy:.4f}")

# Class labels
labels = list(clf.classes_)

# Confusion matrix
confusion = confusion_matrix(
    yte,
    pred_logistic,
    labels=labels
)

confusion_df = pd.DataFrame(
    confusion,
    index=[f"Actual {label}" for label in labels],
    columns=[f"Predicted {label}" for label in labels]
)

print("\nConfusion Matrix:")
print(confusion_df)

# Classification report
print("\nClassification Report:")
print(
    classification_report(
        yte,
        pred_logistic,
        zero_division=0
    )
)
```

Logistic Regression, Decision Tree, Random Forest, and LDA

```python

# ------------------------------------------------------------
# Grid-search models
# ------------------------------------------------------------

models = {
    "Logistic Regression": GridSearchCV(
        estimator=make_pipeline(
            StandardScaler(),
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        ),
        param_grid={
            "logisticregression__C": [0.01, 0.1, 1, 10],
            "logisticregression__class_weight": [
                None,
                "balanced"
            ]
        },
        cv=10,
        scoring="balanced_accuracy",
        n_jobs=2
    ),

    "Decision Tree": GridSearchCV(
        estimator=DecisionTreeClassifier(
            random_state=42
        ),
        param_grid={
            "max_depth": [3, 5, 10, None],
            "min_samples_leaf": [1, 5, 10],
            "class_weight": [None, "balanced"]
        },
        cv=10,
        scoring="balanced_accuracy",
        n_jobs=2
    ),

    "Random Forest": GridSearchCV(
        estimator=RandomForestClassifier(
            n_estimators=500,
            random_state=42,
            n_jobs=2
        ),
        param_grid={
            "max_depth": [None, 5, 10],
            "max_features": [2, 3, 4],
            "min_samples_leaf": [1, 2, 4],
            "class_weight": [None, "balanced"]
        },
        cv=10,
        scoring="balanced_accuracy",
        n_jobs=2
    )
}

# ------------------------------------------------------------
# Store results
# ------------------------------------------------------------

results = []
model_predictions = {}
fitted_models = {}

# ------------------------------------------------------------
# Train and evaluate GridSearchCV models
# ------------------------------------------------------------

for model_name, grid_model in models.items():

    # Train
    grid_model.fit(Xtr, ytr)

    # Predictions
    predictions = grid_model.predict(Xte)

    # Predicted probabilities
    probabilities = grid_model.predict_proba(Xte)

    # Class order
    class_labels = list(grid_model.classes_)

    # Save model and predictions
    fitted_models[model_name] = grid_model
    model_predictions[model_name] = predictions

    # Multiclass AUC
    try:
        multiclass_auc = roc_auc_score(
            yte,
            probabilities,
            multi_class="ovr",
            average="weighted",
            labels=class_labels
        )
    except ValueError:
        multiclass_auc = np.nan

    # Recall for middle-age class
    recall_mid = recall_score(
        yte,
        predictions,
        labels=["mid"],
        average=None,
        zero_division=0
    )[0]

    # F1 for middle-age class
    f1_mid = f1_score(
        yte,
        predictions,
        labels=["mid"],
        average=None,
        zero_division=0
    )[0]

    # Weighted metrics
    weighted_precision = precision_score(
        yte,
        predictions,
        average="weighted",
        zero_division=0
    )

    weighted_recall = recall_score(
        yte,
        predictions,
        average="weighted",
        zero_division=0
    )

    weighted_f1 = f1_score(
        yte,
        predictions,
        average="weighted",
        zero_division=0
    )

    # Add model results
    results.append({
        "Model": model_name,

        "CV Balanced Accuracy": round(
            grid_model.best_score_,
            4
        ),

        "Test Accuracy": round(
            accuracy_score(yte, predictions),
            4
        ),

        "Balanced Accuracy": round(
            balanced_accuracy_score(yte, predictions),
            4
        ),

        "Weighted Precision": round(
            weighted_precision,
            4
        ),

        "Weighted Recall": round(
            weighted_recall,
            4
        ),

        "Weighted F1": round(
            weighted_f1,
            4
        ),

        "Recall Mid": round(
            recall_mid,
            4
        ),

        "F1 Mid": round(
            f1_mid,
            4
        ),

        "Multiclass AUC": (
            round(multiclass_auc, 4)
            if not np.isnan(multiclass_auc)
            else np.nan
        )
    })

    print(f"\n{model_name}")
    print("Best parameters:", grid_model.best_params_)

# ============================================================
# ADD LINEAR DISCRIMINANT ANALYSIS
# ============================================================

lda_model = LinearDiscriminantAnalysis()

# Train LDA
lda_model.fit(Xtr, ytr)

# LDA predictions
pred_lda_final = lda_model.predict(Xte)

# LDA predicted probabilities
proba_lda = lda_model.predict_proba(Xte)

# LDA class order
lda_classes = list(lda_model.classes_)

# LDA cross-validation balanced accuracy
lda_cv_balanced_accuracy = cross_val_score(
    lda_model,
    Xtr,
    ytr,
    cv=10,
    scoring="balanced_accuracy"
).mean()

# LDA multiclass AUC
try:
    lda_auc = roc_auc_score(
        yte,
        proba_lda,
        multi_class="ovr",
        average="weighted",
        labels=lda_classes
    )
except ValueError:
    lda_auc = np.nan

# LDA middle-class recall
lda_recall_mid = recall_score(
    yte,
    pred_lda_final,
    labels=["mid"],
    average=None,
    zero_division=0
)[0]

# LDA middle-class F1
lda_f1_mid = f1_score(
    yte,
    pred_lda_final,
    labels=["mid"],
    average=None,
    zero_division=0
)[0]

# LDA weighted metrics
lda_weighted_precision = precision_score(
    yte,
    pred_lda_final,
    average="weighted",
    zero_division=0
)

lda_weighted_recall = recall_score(
    yte,
    pred_lda_final,
    average="weighted",
    zero_division=0
)

lda_weighted_f1 = f1_score(
    yte,
    pred_lda_final,
    average="weighted",
    zero_division=0
)

# Add LDA to the results
results.append({
    "Model": "Linear Discriminant Analysis",

    "CV Balanced Accuracy": round(
        lda_cv_balanced_accuracy,
        4
    ),

    "Test Accuracy": round(
        accuracy_score(yte, pred_lda_final),
        4
    ),

    "Balanced Accuracy": round(
        balanced_accuracy_score(
            yte,
            pred_lda_final
        ),
        4
    ),

    "Weighted Precision": round(
        lda_weighted_precision,
        4
    ),

    "Weighted Recall": round(
        lda_weighted_recall,
        4
    ),

    "Weighted F1": round(
        lda_weighted_f1,
        4
    ),

    "Recall Mid": round(
        lda_recall_mid,
        4
    ),

    "F1 Mid": round(
        lda_f1_mid,
        4
    ),

    "Multiclass AUC": (
        round(lda_auc, 4)
        if not np.isnan(lda_auc)
        else np.nan
    )
})

# Save LDA for later confusion matrix
model_predictions[
    "Linear Discriminant Analysis"
] = pred_lda_final

fitted_models[
    "Linear Discriminant Analysis"
] = lda_model

print("\nLinear Discriminant Analysis")
print("No hyperparameter grid used.")

# ============================================================
# FINAL COMPARISON TABLE
# ============================================================

classification_results = pd.DataFrame(results)

classification_results = classification_results.sort_values(
    by="Balanced Accuracy",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 140)
print("FINAL CLASSIFICATION MODEL COMPARISON")
print("=" * 140)

print(
    classification_results.to_string(
        index=False
    )
)

print("=" * 140)

# ------------------------------------------------------------
# Majority-class baseline
# ------------------------------------------------------------

majority_class = y_class.value_counts().idxmax()

baseline_accuracy = y_class.value_counts(
    normalize=True
).max()

print(
    f"\nBaseline model always predicts "
    f"'{majority_class}': {baseline_accuracy:.4f}"
)

# ============================================================
# CONFUSION MATRICES AND CLASSIFICATION REPORTS
# ============================================================

for model_name, predictions in model_predictions.items():

    fitted_model = fitted_models[model_name]

    # GridSearchCV models and ordinary LDA both have classes_
    class_labels = list(fitted_model.classes_)

    matrix = confusion_matrix(
        yte,
        predictions,
        labels=class_labels
    )

    matrix_df = pd.DataFrame(
        matrix,
        index=[
            f"Actual {label}"
            for label in class_labels
        ],
        columns=[
            f"Predicted {label}"
            for label in class_labels
        ]
    )

    print("\n" + "-" * 80)
    print(model_name)
    print("-" * 80)

    print("\nConfusion Matrix:")
    print(matrix_df)

    print("\nClassification Report:")
    print(
        classification_report(
            yte,
            predictions,
            labels=class_labels,
            zero_division=0
        )
    )
```
